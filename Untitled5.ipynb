```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('/content/iris.csv') # Added to define df
count = df["species"].value_counts()
plt.bar(count.index,count.values)
plt.xlabel("variety")
plt.ylabel("count")
plt.title("variety count")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
plt.figure(figsize=(8,6))
sns.histplot(df["petal_length"],bins=15,kde=True)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize(8,8))
plt.pie(count,lables=count.index,autopct='1%.1f%%')
plt.title('priportion of each species')
plt.axis('equal')
plt.show()
```

```python
plt.scatter(df['petal_length'], df['petal_width'], color='purple')
plt.title('3. Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.show()
```
